## 4. Arrange the food prices dataset

## 7. Arrange the weather datasets

### 7.1 Define the needed functions

#### 7.1.1 Define the months

```python
months = ["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov","dec"]
month_map = {"jan": 1, "feb": 2, "mar": 3, "apr": 4, "may": 5, "jun": 6, "jul": 7, "aug": 8, "sep": 9, "oct": 10, "nov": 11, "dec": 12}
```

#### 7.1.2 Create a function to convert weather data to long format and map the month names to month numbers

```python
def transform_weather_data(weather_metric_df, metric_name):
    weather_metric_df = weather_metric_df.melt(id_vars="year", value_vars=months, var_name="month_name", value_name=metric_name)
    weather_metric_df["month"] = weather_metric_df["month_name"].map(month_map)
    weather_metric_df = weather_metric_df.drop(columns=["month_name"])
    weather_metric_df = weather_metric_df[["year", "month", metric_name]]
    return weather_metric_df
```

### 7.2 Get transformed weather datasets

```python
air_frost_df = transform_weather_data(days_of_air_frost_df, "air_frost")
sunshine_df = transform_weather_data(hours_of_sunshine_df, "sunshine")
temperature_df = transform_weather_data(mean_temperature_df, "temperature")
rainfall_df = transform_weather_data(rainfall_figures_df, "rainfall")
```

## 9. Get the correlation matrix

### 9.1 Get the correlation matrix of the merged dataset

```python
feature_list = ["food_price", "fuel_price", "air_frost", "sunshine", "temperature", "rainfall"]
corr_matrix = merged_food_fuel_weather_df[feature_list].corr()
seaborn.heatmap(corr_matrix, annot=True, cmap="coolwarm", center=0)
pyplot.show()
```

Temperature has the strongest positive correlation with food price. These tend to appear when temperatures are higher and are often more expensive than winter vegetables. So, it is likely that the correlation is due to the seasonal nature of food prices rather than a direct causal relationship between temperature and food prices.

Air frost has the strongest negative correlation with food prices. This is likely because air frost is more common in the winter months when food prices are lower.

### 9.2 Get the correlation bar chart

```python
correlation_with_food_price = merged_food_fuel_weather_df.corr(numeric_only=True)["food_price"]

correlation_with_food_price = correlation_with_food_price.drop(["year", "month", "food_price"])

correlation_with_food_price.sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

pyplot.xlabel("Correlation Coefficient")
pyplot.title("Correlation with Food Price")
pyplot.show()
```

### 9.3 Plot the relationship between food prices and temperature

#### 9.3.1 Define a function to plot the relationship between two variables

```python
def plot_correlation(x, y, x_label, y_label):
    pyplot.figure(figsize=(8, 6))
    pyplot.scatter(x, y, alpha=0.7)
    m, c = numpy.polyfit(x, y, 1)
    pyplot.plot(
        x,
        m * x + c,
        color="red",
        linewidth=2,
        label=f"y = {m:.3f}x + {c:.3f}"
    )
    corr = x.corr(y)
    pyplot.xlabel(x_label)
    pyplot.ylabel(y_label)
    pyplot.title(f"{y_label} vs {x_label}\nCorrelation = {corr:.3f}")
    pyplot.legend()
    pyplot.grid(alpha=0.3)
    pyplot.show()
```

Define the food price as the y axis

```python
y = merged_food_fuel_weather_df["food_price"]
```

#### 9.3.2 Plot the relationship between food prices and temperature

```python
x = merged_food_fuel_weather_df["temperature"]
plot_correlation(x, y, "Temperature", "Food Price")
```

#### 9.3.3 Plot the relationship between food prices and air frost

```python
x = merged_food_fuel_weather_df["air_frost"]
plot_correlation(x, y, "Air Frost", "Food Price")
```

## 10. Get the monthly average food prices and temperature

### 10.1 Get the monthly average food prices and temperature

```python
food_temp_monthly_pattern_df = (
    merged_food_fuel_weather_df.groupby("month")
    .agg({
        "food_price": "mean",
        "temperature": "mean"
    })
    .reset_index()
)

food_temp_monthly_pattern_df = food_temp_monthly_pattern_df[["month", "temperature", "food_price"]]
food_temp_monthly_pattern_df.head(12)
```

### 10.2 Plot the monthly average food prices and temperature

```python
fig, ax1 = pyplot.subplots(figsize=(10, 6))

ax1.plot(
    food_temp_monthly_pattern_df["month"],
    food_temp_monthly_pattern_df["food_price"],
    color="blue",
    marker="o",
    label="Food Price"
)

ax1.set_ylabel("Food Price", color="blue")

ax2 = ax1.twinx()

ax2.plot(
    food_temp_monthly_pattern_df["month"],
    food_temp_monthly_pattern_df["temperature"],
    color="red",
    marker="o",
    label="Temperature"
)

ax2.set_ylabel("Temperature", color="red")
pyplot.title("Average Monthly Food Price vs Temperature")
pyplot.show()
```

## 11. Plot the seasonal pattern of food prices and temperature

### 11.1 Define the seasons based on the month

```python
season_map = {
    12: "Winter",
    1: "Winter",
    2: "Winter",
    3: "Spring",
    4: "Spring",
    5: "Spring",
    6: "Summer",
    7: "Summer",
    8: "Summer",
    9: "Autumn",
    10: "Autumn",
    11: "Autumn"
}

food_temp_monthly_pattern_df["season"] = food_temp_monthly_pattern_df["month"].map(season_map)
food_temp_monthly_pattern_df = food_temp_monthly_pattern_df[["month", "season", "temperature", "food_price"]]
food_temp_monthly_pattern_df.head(12)
```

### 11.2 Get the seasonal pattern of food prices and temperature

```python
food_temp_seasonal_pattern_df = (
    food_temp_monthly_pattern_df.groupby("season")
    .agg({
        "food_price": "mean",
        "temperature": "mean"
    })
    .round(2)
)

food_temp_seasonal_pattern_df.head(4)
```

### 11.3 Plot the seasonal pattern of food prices and temperature bar chart

```python
season_order = ["Winter", "Spring", "Summer", "Autumn"]
fig, axes = pyplot.subplots(1, 2, figsize=(12, 5))

seaborn.barplot(
    data=food_temp_seasonal_pattern_df,
    x="season",
    y="food_price",
    order=season_order,
    ax=axes[0]
)

axes[0].set_title("Average Food Price by Season")

seaborn.barplot(
    data=food_temp_seasonal_pattern_df,
    x="season",
    y="temperature",
    order=season_order,
    ax=axes[1]
)

axes[1].set_title("Average Temperature by Season")
pyplot.tight_layout()
pyplot.show()
```
